```python
import torch
import gc

def clear_cuda_memory():
    gc.collect()  # Сборщик мусора Python
    torch.cuda.empty_cache()  # Очистка кэша CUDA
    # Дополнительно: сброс статистики памяти (необязательно)
    torch.cuda.reset_peak_memory_stats()  
```

```python
def eval_model_on_env(model, env, test_size=1_000, batch_size=64):

    model.eval()

    torch.cuda.empty_cache()  # Чистим кэш CUDA

    res = []
    remind_n = test_size

    while remind_n > 0:
        print(remind_n, end="\r")

        bs = min(batch_size, remind_n)

        td = env.reset(batch_size=bs).to(model.device)

        out = model(td)

        res += out["reward"].cpu().tolist()  # Переносим на CPU перед конвертацией
        del td, out  # Удаляем ссылки
        torch.cuda.empty_cache()  # Чистим кэш CUDA

        remind_n = max(0, remind_n-batch_size)


    return -1* sum(res)/test_size
```

```python
clear_cuda_memory()
eval_model_on_env(model, env, test_size=1_000, batch_size=64)
```

```python
class FJSPEnvAssamblyConstarints(FJSPEnv):
    def __init__(self, generator = None, generator_params = ..., mask_no_ops = True, check_mask = False, stepwise_reward = False, **kwargs):
        self.max_ops_processed = generator_params.pop("max_ops_processed")

        super().__init__(generator, generator_params, mask_no_ops, check_mask, stepwise_reward, **kwargs)

    def _reset(self, td = None, batch_size=None):
        td = super()._reset(td, batch_size)

        ma_ops_processed_left = torch.ones_like(td["busy_until"]) * self.max_ops_processed

        td["ma_ops_processed_left"] = ma_ops_processed_left

        return td.to(self.device)
    
    def _step(self, td):

        td = td.to(self.device)

        n_batches, n_jobs = td["end_op_per_job"].shape

        ma_ops_processed_left = td["ma_ops_processed_left"]

        n_machines = ma_ops_processed_left.size(1)

        machines = (td["action"] - 1) % n_machines

        ma_ops_processed_left[torch.arange(ma_ops_processed_left.size(0), device=self.device)[td["action"] > 0], machines[td["action"] > 0]] -= 1

        assert (ma_ops_processed_left >= 0).all()

        td = super()._step(td)

        td["ma_ops_processed_left"] = ma_ops_processed_left

        availible_machines_mask = (ma_ops_processed_left > 0)

        next_op_expanded = td["next_op"].unsqueeze(1).expand(-1, n_machines, -1)

        needed_machines = torch.gather(td["ops_ma_adj"], 2, next_op_expanded).sum(dim=-1)

        can_do_op = (availible_machines_mask * (needed_machines > 0)).any(dim=1, keepdim=True)

        new_mask = torch.concat([torch.tensor([True], device=self.device).bool().repeat(n_batches, 1), availible_machines_mask.repeat(1, n_jobs)], dim=1)

        new_mask = td["action_mask"] * new_mask

        all_false_rows = ~new_mask.any(dim=1)

        new_mask[all_false_rows, 0] = True

        td["action_mask"] = new_mask

        if (~can_do_op+td["done"]).any():

            td["time"][(~can_do_op+td["done"]).squeeze(1)] += 100*(~td["done"].squeeze(1)[(~can_do_op+td["done"]).squeeze(1)])

            td["done"][~can_do_op + td["done"]] = True

        return td

```
